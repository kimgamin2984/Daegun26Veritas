```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv('/workspaces/Daegun26Veritas/data/10727data.csv')

years_to_process = [2019, 2020, 2021, 2022, 2023, 2024]

# Initialize data lists for drug categories
at_data = [] # 항악성종양제
pa_data = [] # 해열·진통·소염제
ct_data = [] # X선조영제
vc_data = [] # 백신류

# Define drug categories and their possible naming variations in the dataset
drug_categories_map = {
    '항악성종양제': {'list': at_data, 'names': ['항악성종양제']},
    '해열·진통·소염제': {'list': pa_data, 'names': ['해열·진통·소염제', '해열 진통소염제']},
    'X선조영제': {'list': ct_data, 'names': ['X선조영제']},
    '백신류': {'list': vc_data, 'names': ['백신류']}
}

for year in years_to_process:
    symptom_col = f'연도별증상({year})'
    count_col = f'연도별보고건수({year})'

    if symptom_col in df.columns and count_col in df.columns:
        for drug_category_name, drug_info in drug_categories_map.items():
            data_list = drug_info['list']
            names_to_match = drug_info['names']

            # Filter rows where the symptom column value (after stripping) is in our list of names_to_match
            filtered_df_year = df[df[symptom_col].fillna('').str.strip().isin(names_to_match)]

            if not filtered_df_year.empty:
                value = filtered_df_year[count_col].iloc[0]
                data_list.append(value)
            else:
                data_list.append(None)
    else:
        # If columns for a specific year are missing, append None for all drugs for that year
        for drug_category_name, drug_info in drug_categories_map.items():
            drug_info['list'].append(None)

# Now assign the extracted data to the original variable names for plotting
years = years_to_process
at = at_data
pa = pa_data
ct = ct_data
vc = vc_data

plt.figure(figsize=(10,6))

plt.plot(years, at, marker='o', label='항악성종양제')
plt.plot(years, pa, marker='o', label='해열·진통·소염제')
plt.plot(years, ct, marker='o', label='X선조영제')
plt.plot(years, vc, marker='o', label='백신류')

plt.title('2019~2024 의약품 부작용 신고 건수 변화')
plt.xlabel('연도')
plt.ylabel('신고 건수')
plt.legend()

plt.show()
```

```python
years = [2019, 2020, 2021, 2022, 2023, 2024]

at = [31020, 34951, 33064, 28165, 26145, 31748]
pa = [38591, 37443, 31768, 32573, 34644, 31228]
ct = [20377, 20232, 19547, 26323, 27604, 27859]
vc = [None, None, 300425, 66579, 5965, 5301]

plt.figure(figsize=(10,6))

plt.plot(years, at, marker='o', label='항악성종양제')
plt.plot(years, pa, marker='o', label='해열·진통·소염제')
plt.plot(years, ct, marker='o', label='X선조영제')
plt.plot(years, vc, marker='o', label='백신류')

plt.title('2019~2024 의약품 부작용 신고 건수 변화')
plt.xlabel('연도')
plt.ylabel('신고 건수')
plt.legend()

plt.show()

drugs = {
    "항악성종양제": at,
    "해열·진통·소염제": pa,
    "X선조영제": ct
}

print("\n=== 분석 결과 ===")

# 증감률 계산
rates = {}

for name, data in drugs.items():
    rate = (data[-1] - data[0]) / data[0] * 100
    rates[name] = rate

# 가장 증가한 효능군
max_drug = max(rates, key=rates.get)

# 가장 감소한 효능군
min_drug = min(rates, key=rates.get)

print(f"가장 큰 증가율 : {max_drug} ({rates[max_drug]:.1f}%)")
print(f"가장 큰 감소율 : {min_drug} ({rates[min_drug]:.1f}%)")

# 전체 데이터 중 최대값 찾기
all_data = {
    "항악성종양제": at,
    "해열·진통·소염제": pa,
    "X선조영제": ct,
    "백신류": vc
}

max_value = 0
max_drug_name = ""
max_year = ""

for name, values in all_data.items():
    for year, value in zip(years, values):

        if value is not None and value > max_value:
            max_value = value
            max_drug_name = name
            max_year = year

print(f"최고 신고 건수 : {max_drug_name} ({max_year}년, {max_value:,}건)")
```
